# Antenna Panel

```python
import numpy as np
import scipy.io

import neoradium as nr
```

```python
# First, we create an antenna element template. The 'AntennaPanel' class uses this template
# to create the panel elements.
elementTemplate = nr.AntennaElement(beamWidth=[65,65], maxAttenuation=30)

# Next, we create the antenna panel using the element template.
# Note that the spacing between elements is 0.5 λ by default.
panel = nr.AntennaPanel([4,4], elements=elementTemplate, polarization="x")

# The 'showElements' method visualizes the antenna panel by displaying all its elements.
panel.showElements(zeroTicks=True)
```

```python
# Depending on the input parameters, the 'drawRadiation' method can generate different types of plots.
# Here, we plot the antenna panel field values in the horizontal plane at zero elevation.
ax = panel.drawRadiation(theta=90, radiationType="Field", normalize=False)
```

```python
# Here, the 'drawRadiation' method is used to plot the field values in the vertical plane at an azimuth angle of 0°.
ax = panel.drawRadiation(phi=0, radiationType="Field", normalize=False)
```

```python
# Here, the 'drawRadiation' method is used to generate a 3D plot of directivity.
ax = panel.drawRadiation(radiationType="Directivity", normalize=True, viewAngles=(60,10))
```
